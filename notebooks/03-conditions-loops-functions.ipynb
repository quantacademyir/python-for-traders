## ۱. تکرار را به کامپیوتر بسپار

```python
closes = [98500, 99200, 97800, 99900, 101200, 100400, 102000]
```

## ۲. حلقه for

```python
returns = []

for i in range(1, len(closes)):
    r = closes[i] / closes[i - 1] - 1
    returns.append(r)

for r in returns:
    print(f"{r:.2%}")
```

## ۳. شرط if

```python
up_days = 0

for r in returns:
    if r > 0:
        up_days += 1

print(up_days, "از", len(returns))
```

```python
for r in returns:
    if r > 0:
        print("صعودی")
    else:
        print("نزولی یا بدون تغییر")
```

## ۴. تابع

```python
def simple_return(old, new):
    return new / old - 1

print(f"{simple_return(98500, 99200):.4%}")
```

```python
def daily_returns(prices):
    out = []
    for i in range(1, len(prices)):
        out.append(simple_return(prices[i - 1], prices[i]))
    return out

print([round(r, 4) for r in daily_returns(closes)])
```

## ۵. بیشترین افت از اوج (Drawdown)

```python
peak = closes[0]

for p in closes:
    if p > peak:
        peak = p
    print(p, peak, f"{p / peak - 1:.2%}")
```

```python
def max_drawdown(prices):
    peak = prices[0]
    worst = 0.0
    for p in prices:
        if p > peak:
            peak = p
        dd = p / peak - 1
        if dd < worst:
            worst = dd
    return worst

print(f"{max_drawdown(closes):.2%}")
print(f"{max_drawdown([100, 120, 90, 95, 130, 104]):.2%}")
```
