## ۱. میانگین پنجره‌ای

```python
import numpy as np
import pandas as pd

URL = "https://raw.githubusercontent.com/quantacademyir/python-for-traders/main/data/btc_daily_sample.csv"
df = pd.read_csv(URL, parse_dates=["date"], index_col="date")
close = df["close"]
```

## ۲. rolling

```python
ma5 = close.rolling(5).mean()
print(ma5.head(6).round(1))
```

```python
print(close.iloc[:5].mean())
```

```python
print(close.rolling(20, min_periods=1).mean().head(3).round(1))
```

## ۳. میانگین ۲۰ و ۵۰ روزه

```python
df["ma20"] = close.rolling(20).mean()
df["ma50"] = close.rolling(50).mean()

print(df[["close", "ma20", "ma50"]].iloc[18:22].round(1))
print(df["ma20"].isna().sum(), df["ma50"].isna().sum())
```

## ۴. قیمت بالای میانگین است یا زیرش؟

```python
df["above"] = df["close"] > df["ma20"]
print(df["above"].sum())
print(f'{df["above"].iloc[19:].mean():.1%}')
```

```python
df["dist"] = df["close"] / df["ma20"] - 1
print(df["dist"].describe().round(4))
```

## ۵. تقاطع دو میانگین و یک تله

```python
fast_above = df["ma20"] > df["ma50"]
changed = fast_above != fast_above.shift(1)
print(changed.sum())
```

```python
print(fast_above.iloc[47:52].tolist())
print(df["ma50"].notna().iloc[47:52].tolist())
```

```python
valid = df["ma50"].notna()
real = changed & valid & valid.shift(1, fill_value=False)
print(real.sum())

cross = df.loc[real, ["close", "ma20", "ma50"]].round(1)
cross["up"] = fast_above[real]
print(cross)
```

## ۶. نوسان متحرک

```python
ret = close.pct_change()
df["vol20"] = ret.rolling(20).std()
print(df["vol20"].dropna().describe().round(4))
print(df["vol20"].idxmax().date(), round(df["vol20"].max(), 4))
```
