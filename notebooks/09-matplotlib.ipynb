## ۲. آماده‌سازی

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

URL = "https://raw.githubusercontent.com/quantacademyir/python-for-traders/main/data/btc_daily_sample.csv"
df = pd.read_csv(URL, parse_dates=["date"], index_col="date")

df["ret"] = df["close"].pct_change()
df["ma20"] = df["close"].rolling(20).mean()
df["ma50"] = df["close"].rolling(50).mean()
```

## ۳. اسکلت یک نمودار

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df.index, df["close"], label="Close")
ax.set_title("Simulated BTC daily close")
ax.set_ylabel("Price (USDT)")
plt.show()
```

## ۴. نمودار اول: قیمت و میانگین‌ها

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df.index, df["close"], label="Close", linewidth=1.4)
ax.plot(df.index, df["ma20"], label="MA20", linewidth=1.1)
ax.plot(df.index, df["ma50"], label="MA50", linewidth=1.1)
ax.set_title("Simulated BTC daily close with moving averages")
ax.set_ylabel("Price (USDT)")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```

## ۵. نمودار دوم: توزیع بازده‌ها

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(df["ret"].dropna() * 100, bins=40, edgecolor="white")
ax.axvline(0, color="black", linewidth=1)
ax.set_title("Distribution of daily returns")
ax.set_xlabel("Daily return (%)")
ax.set_ylabel("Number of days")
plt.show()
```

## ۶. نمودار سوم: افت از اوج

```python
dd = df["close"] / df["close"].cummax() - 1
print(f"{dd.min():.2%} {dd.idxmin().date()}")
```

```python
fig, ax = plt.subplots(figsize=(10, 3))
ax.fill_between(dd.index, dd * 100, 0, alpha=0.5)
ax.set_title("Drawdown from running peak")
ax.set_ylabel("Drawdown (%)")
ax.grid(alpha=0.3)
plt.show()
```

## ۷. چند نمودار در یک بوم

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 9))
axes[0].plot(df.index, df["close"])
axes[1].hist(df["ret"].dropna() * 100, bins=40)
axes[2].fill_between(dd.index, dd * 100, 0, alpha=0.5)
fig.tight_layout()
fig.savefig("report.png", dpi=130, bbox_inches="tight")
```
