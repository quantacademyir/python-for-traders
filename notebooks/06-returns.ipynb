## ۱. چرا بازده و نه قیمت؟

```python
import numpy as np
import pandas as pd

URL = "https://raw.githubusercontent.com/quantacademyir/python-for-traders/main/data/btc_daily_sample.csv"
df = pd.read_csv(URL, parse_dates=["date"], index_col="date")
close = df["close"]
```

## ۲. بازده ساده با pct_change

```python
ret = close.pct_change()
print(ret.head(4))
```

## ۳. shift: دیروز در سطر امروز

```python
print(close.shift(1).head(3))
```

```python
manual = close / close.shift(1) - 1
print(manual.equals(ret))
```

## ۴. بازده لگاریتمی

```python
logret = np.log(close / close.shift(1))
print(pd.DataFrame({"simple": ret, "log": logret}).head(4).round(5))
```

## ۵. جمع کردن بازده‌ها

```python
total = close.iloc[-1] / close.iloc[0] - 1
print(f"{total:.2%}")
```

```python
print(f"{ret.sum():.4f}")          # جمع بازده ساده
print(f"{logret.sum():.4f}")       # جمع بازده لگاریتمی
print(f"{np.expm1(logret.sum()):.2%}")
```

```python
cum = (1 + ret.fillna(0)).cumprod()
print(cum.tail(2).round(4))
```

## ۶. چند آمار از بازده‌ها

```python
print(f"{ret.max():.2%} {ret.idxmax().date()}")
print(f"{ret.min():.2%} {ret.idxmin().date()}")
print(f"{ret.mean():.4%} {ret.std():.4%}")

r = ret.dropna()
print((r > 0).sum(), (r < 0).sum(), f"{(r > 0).mean():.1%}")
```

## ۷. یک پرسش جالب

```python
print(ret.corr(ret.shift(1)))
```
