## ۳. لیست قیمت‌ها

```python
closes = [98500, 99200, 97800, 99900, 101200, 100400, 102000]
```

```python
print(len(closes))      # تعداد عضوها
print(closes[0])        # اولین عضو
print(closes[-1])       # آخرین عضو
print(closes[2:5])      # از اندیس ۲ تا قبل از ۵
```

```python
print(min(closes))
print(max(closes))
print(sum(closes) / len(closes))
```

## ۴. بازده روزانه با اندیس

```python
r1 = closes[1] / closes[0] - 1
r2 = closes[2] / closes[1] - 1

print(f"{r1:.4%}")
print(f"{r2:.4%}")
```

```python
total = closes[-1] / closes[0] - 1
print(f"{total:.2%}")
```

```python
closes.append(103500)
print(len(closes), closes[-1])
```

## ۵. دیکشنری: یک کندل

```python
candle = {"open": 99900, "high": 101800, "low": 99500, "close": 101200}

print(candle["high"])
```

```python
price_range = candle["high"] - candle["low"]
body = candle["close"] / candle["open"] - 1

print(price_range)
print(f"{price_range / candle['open']:.2%}")
print(f"{body:.2%}")
```

```python
candle["volume"] = 1250.5
print(candle)
```
