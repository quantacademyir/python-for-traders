## ۴. اولین سلول

```python
buy = 99000
sell = 101000

profit = sell - buy
ret = profit / buy

print(profit)
print(ret)
```

```python
print(f"{ret:.2%}")
```

## ۵. کارمزد را هم اضافه کن

```python
fee_rate = 0.001

fees = buy * fee_rate + sell * fee_rate
net_profit = profit - fees

print(fees)
print(net_profit)
print(f"{net_profit / buy:.2%}")
```
