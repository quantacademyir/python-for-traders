## ۱. جدول به‌جای لیست

```python
import pandas as pd

URL = "https://raw.githubusercontent.com/quantacademyir/python-for-traders/main/data/btc_daily_sample.csv"
df = pd.read_csv(URL)
```

## ۲. نگاه اول

```python
print(df.shape)
print(list(df.columns))
print(df.dtypes)
```

```python
print(df.describe().round(1))
```

## ۳. تاریخ به‌عنوان اندیس

```python
df["date"] = pd.to_datetime(df["date"])
df = df.set_index("date")
print(df.head(3))
```

## ۴. انتخاب ستون و سطر

```python
print(df["close"].head(3))
```

```python
print(df.loc["2025-01-03"])
```

```python
print(df.loc["2025-01-05":"2025-01-08", ["open", "close"]])
```

```python
print(df.iloc[0]["close"], df.iloc[-1]["close"])
```

## ۵. فیلتر کردن

```python
big = df[df["close"] > 110000]
print(len(big))
print(big.index[0].date(), big.index[-1].date())
```

```python
df[(df["close"] > 110000) & (df["volume"] > 2000)]
```

## ۶. مرتب کردن

```python
top = df.sort_values("volume", ascending=False).head(3)
print(top[["close", "volume"]])
```

## ۷. ساخت ستون جدید

```python
df["range"] = df["high"] - df["low"]
df["range_pct"] = df["range"] / df["open"]
print(df[["range", "range_pct"]].head(3).round(4))
```

```python
print(df["close"].mean(), df["close"].min(), df["close"].idxmin())
```
