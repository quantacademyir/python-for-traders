## ۴. اختیاری: گرفتن داده با API

```python
rows = [
    [1735689600000, "93576.00", "95151.14", "92888.00", "94591.79", "10373.1", 1735775999999, "...", 1, "...", "...", "0"],
    [1735776000000, "94591.78", "98976.20", "94319.46", "96984.79", "21406.3", 1735862399999, "...", 1, "...", "...", "0"],
]
```

```python
def klines_to_df(rows):
    cols = ["open_time", "open", "high", "low", "close", "volume",
            "close_time", "quote_volume", "trades", "taker_base", "taker_quote", "ignore"]
    df = pd.DataFrame(rows, columns=cols)
    df["date"] = pd.to_datetime(df["open_time"], unit="ms", utc=True)
    df = df.set_index("date")
    for c in ["open", "high", "low", "close", "volume"]:
        df[c] = df[c].astype(float)
    return df[["open", "high", "low", "close", "volume"]]

print(klines_to_df(rows))
```
